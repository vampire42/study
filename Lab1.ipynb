### 1. Арифметические операторы

| Оператор | Название | Пример | Результат |
|----------|----------|--------|-----------|
| `+` | Сложение | `5 + 3` | `8` |
| `-` | Вычитание | `5 - 3` | `2` |
| `*` | Умножение | `5 * 3` | `15` |
| `/` | Деление | `5 / 2` | `2.5` (всегда float!) |
| `//` | Целочисленное деление | `5 // 2` | `2` |
| `%` | Остаток от деления | `5 % 2` | `1` |
| `**` | Возведение в степень | `2 ** 3` | `8` |

### 2. Ввод/вывод данных

```python
# Вывод
print("Текст")
print(f"Значение: {x}")  # f-строка

# Ввод
name = input("Введите имя: ")  # Всегда возвращает str!
```

### РАЗДЕЛ 1: АРИФМЕТИЧЕСКИЕ ОПЕРАЦИИ (25 мин)

````python
# Пример 1.1: Базовые операции
a = 25
b = 4

print(f"a = {a}, b = {b}")
print(f"a + b = {a + b}")      # Сложение
print(f"a - b = {a - b}")      # Вычитание
print(f"a * b = {a * b}")      # Умножение
print(f"a / b = {a / b}")      # Деление (всегда float!)
print(f"a // b = {a // b}")    # Целочисленное деление
print(f"a % b = {a % b}")      # Остаток от деления
print(f"a ** b = {a ** b}")    # Возведение в степень
````

````python
# Пример 1.2: Особенности деления

# Деление всегда возвращает float
print(f"10 / 2 = {10 / 2}")        # 5.0 (не 5!)
print(f"type(10 / 2) = {type(10 / 2)}")  # <class 'float'>

# Целочисленное деление с отрицательными числами
print(f"7 // 2 = {7 // 2}")        # 3
print(f"-8 // 2 = {-8 // 2}")      # -4 (округление ВНИЗ!)

# Остаток от деления
print(f"7 % 3 = {7 % 3}")          # 1
print(f"-8 % 2 = {-8 % 2}")        # 2 (знак делителя!)
````

````python
# Пример 1.3: Приоритет операций

# Приоритет операций
x = 2 + 3 * 4
print(f"2 + 3 * 4 = {x}")          # 14 (не 20!)

# Использование скобок
y = (2 + 3) * 4
print(f"(2 + 3) * 4 = {y}")        # 20

# Возведение в степень (правоассоциативно)
z = 2 ** 3 ** 2
print(3**2)
print(2** (3**2))
print(2**3)
print((2**3)**2)
print(f"2 ** 3 ** 2 = {z}")        # 512 (2 ** 9)
print(f"(2 ** 3) ** 2 = {(2 ** 3) ** 2}")  # 64
````

````python
# Пример 1.4: Практическое применение

# Задача: Расчёт стоимости покупки
price = 599.90      # цена за товар
quantity = 5        # количество
discount = 0.15     # скидка 15%

subtotal = price * quantity
discount_amount = subtotal * discount
total = subtotal - discount_amount

print("=" * 30)
print("ЧЕК ПОКУПКИ")
print("=" * 30)
print(f"Цена за штуку: {price:.2f} ₽")
print(f"Количество: {quantity}")
print(f"Подытог: {subtotal:.2f} ₽")
print(f"Скидка (15%): {discount_amount:.2f} ₽")
print(f"ИТОГО: {total:.2f} ₽")
print("=" * 30)
````

### РАЗДЕЛ 3: РАБОТА С ТИПАМИ ДАННЫХ (20 мин)

````python
# Пример 3.1: Основные типы данных

# Целые числа (int)
age = 20
count = -5
large_number = 10**100  # Python поддерживает большие числа!

print(f"age = {age}, type: {type(age)}")
print(f"count = {count}, type: {type(count)}")

# Числа с плавающей точкой (float)
pi = 3.14159
price = 599.90
scientific = 6.022e23  # Число Авогадро

print(f"pi = {pi}, type: {type(pi)}")
print(f"scientific = {scientific}, type: {type(scientific)}")

# Строки (str)
name = "Алексей"
university = 'МГТУ им. Баумана'
multiline = """Это
многострочная
строка"""

print(f"name = {name}, type: {type(name)}")
print(f"Длина имени: {len(name)}")

# Булевы значения (bool)
is_student = True
is_graduated = False

print(f"is_student = {is_student}, type: {type(is_student)}")
print(f"True + True = {True + True}")  # 2 (bool - подкласс int!)
````

````python
# Пример 3.2: Преобразование типов

# str → int/float
num_str = "123"
num_int = int(num_str)
print(f"int('123') = {num_int}, type: {type(num_int)}")

float_str = "3.14"
float_num = float(float_str)
print(f"float('3.14') = {float_num}, type: {type(float_num)}")

# int/float → str
number = 42
str_number = str(number)
print(f"str(42) = '{str_number}', type: {type(str_number)}")

# В bool
print(f"bool(0) = {bool(0)}")        # False
print(f"bool(1) = {bool(1)}")        # True
print(f"bool('') = {bool('')}")      # False (пустая строка)
print(f"bool('0') = {bool('0')}")    # True (непустая строка!)
````

````python
# Пример 3.3: Проверка типов

# Функция type()
x = 42
print(f"type(x) = {type(x)}")

# Функция isinstance()
print(f"isinstance(x, int) = {isinstance(x, int)}")    # True
print(f"isinstance(x, float) = {isinstance(x, float)}") # False

# Проверка нескольких типов
y = 3.14
print(f"isinstance(y, (int, float)) = {isinstance(y, (int, float))}")  # True
````

````python
# ВАШ КОД ЗДЕСЬ: Задание 3.2
# Проверка, является ли число целым  (if-else)

number = input("Введите число: ")

try:
    number = float(number)

    if number.is_integer():
        print(f"Поздравляем, число {int(number)} целое")
    else:
        print(f"К сожалению, число {number} не является целым")

except ValueError:
    print("Ошибка! Неверный формат данных.")
````

````python
# ВАШ КОД ЗДЕСЬ: Задание 3.3
# Калькулятор ИМТ if-elif-else


try:
    hieght = input("Введите свой рост в сантиметрах: ")
    weight = input("Введите свой вес в килограммах: ")

    hieght = int(hieght) / 100
    weight = int(weight)
    if weight or hieght <= 0 :
        print("Ну Вы шутник! Здесь не может быть отрицательных значений")
    else:
        if hieght < 350:
            body_mass_index = weight / (hieght ** 2)

            if body_mass_index < 18.5:
                print(f"Ваш индекс массы тела: {body_mass_index:.1f}. У вас недостаточный вес.")
            elif body_mass_index > 18.5 and body_mass_index < 24.9:
                print(f"Ваш индекс массы тела: {body_mass_index:.1f}. У вас нормальный вес.")
            elif body_mass_index > 25 and body_mass_index < 29.9:
                    print(f"Ваш индекс массы тела: {body_mass_index:.1f}. У вас избыточный вес.")
            else:
                    print(f"Ваш индекс массы тела: {body_mass_index:.1f}. У вас ожирение.")
        else:
             print("Несуществующий рост")
except ValueError:
    print("Неверный формат данных")
````

````python
# ВАШ КОД ЗДЕСЬ: Задание 3.4
# Демонстрация проблемы точности float  -   math.isclose(a, b)

from math import isclose

try:
    number = float(input("Введите число с плавающей точкой: "))

    compare_number = number * 3
    compare_number = compare_number / 3

    print(f"Ваше число: {compare_number}")

    print("===Прямое сравнение===")

    if compare_number == number:
        print("равно")
    else:
            print("Числа не равны!")

    print("===Сравнение через isclose===")

    if isclose(number, compare_number):
        print("Числа равны!")
    else:
        print("Числа не равны!")

except ValueError:
    print("Ошибка! Вы ввели число в неправильном формате. Запишите дробную часть через точку")
````

Напишите программу, которая рассчитывает стоимость коммунальных услуг:

   - Показания счётчика электроэнергии (начальные и конечные)
   - Тариф за 1 кВт·ч (в рублях)

2. Вычислите:
   - Расход воды (куб.м)
   - Стоимость воды
   - Расход электроэнергии (кВт·ч)
   - Стоимость электроэнергии

3. Выведите красивый чек с форматированием

````python
# ВАШ КОД ЗДЕСЬ: Итоговое задание
# Калькулятор коммунальных платежей

print("=" * 40)
print("   КАЛЬКУЛЯТОР КОММУНАЛЬНЫХ ПЛАТЕЖЕЙ")
print("=" * 40)

try:
    s_water_meter = float(input("Введите начальные показатели счетчика воды: "))
    f_water_meter = float(input("Введите конечные показатели счетчика воды: "))

    if s_water_meter > f_water_meter:
        print("Ошибка! Начальное число не может быть больше конечного!")

    elif s_water_meter < 0 or f_water_meter < 0:
        print("Ошибка! Показатели счетчика не могут быть отрицательными!")

    else:
        water_tarif = int(input("Введите тариф по воде: "))
        if water_tarif <= 0:
            print("Ошибка. Тариф не может принимать таких значений!")

        else:
            s_electric_meter = float(input("Введите начальные показатели счетчика электричества: "))
            f_electric_meter = float(input("Введите конечные показатели счетчика электричества: "))

            if s_electric_meter > f_electric_meter:
                print("Ошибка! Начальное число не может быть больше конечного!")

            elif s_electric_meter < 0 or f_electric_meter < 0:
                    print("Ошибка! Показатели счетчика не могут быть отрицательными!")

            else:
                electric_tarif = int(input("Введите тариф по электричеству: "))

                if water_tarif <= 0:
                            print("Ошибка. Тариф не может принимать таких значений!")

                else:
                    water_spend = f_water_meter - s_water_meter #сколько воды затрачено
                    water_amount = water_spend * water_tarif #высчитываем сумму к оплате

                    electric_spent = f_electric_meter - s_electric_meter #сколько электричества затрачено
                    electric_amount = electric_spent * electric_tarif #высчитываем сумму к оплате

                    total_amount = water_amount + electric_amount #итоговая сумма

                    print(f"{'Требуется к оплате счетов за воду:':>15} {water_amount:>20}")
                    print(f"{'Требуется к оплате счетов за электричество:':>15} {electric_amount:>11}")

                    print(f"ИТОГО к оплате: {total_amount:>39}")

except ValueError:
    print("Ошибка. Введены неверные значения!")
````
